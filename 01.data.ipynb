```python
import numpy as np
import pandas as pd 
#yüklediğimiz veri setini okuyoruz
df=pd.read_csv("cumulative.csv")

df.head()

```

```python
#hedef sütunumuz olan koi_disposition sütunundaki hedef değişkenlerin sayılarını buluyoruz.
df["koi_disposition"].value_counts()
#aday olarak işaretlenmiş CANDIDATE değişkenlerini çıkartıp df_filtered isimli yeni bir tablo oluşturuyoruz.
df_filtered=df[df["koi_disposition"]!='CANDIDATE']
df_filtered.head()
```

```python
df_filtered["koi_disposition"].value_counts()
```

```python
#mapping değişkeni ile bir python sözlüğü oluşturup alttaki güncelleme fonksiyonu olan map() fonksiyonuna parametre olarak veriyoruz.
mapping={"FALSE POSITIVE":0,"CONFIRMED":1}
df_filtered['koi_disposition']=df_filtered['koi_disposition'].map(mapping)
df_filtered.head()
```

```python

df_filtered.isnull().sum()
```

```python
columns_to_drop=['rowid', 'kepid', 'kepoi_name', 'kepler_name', 'koi_score','koi_pdisposition']
df_filtered=df_filtered.drop(columns_to_drop,axis=1)
```

```python
# 1. İçinde 'err' geçen kolonları otomatik bul ve bir listeye at
err_columns = [col for col in df_filtered.columns if 'err' in col]

# 2. Bulunan bu kolonları veri setinden (sütun ekseninde, axis=1) düşür
df_filtered = df_filtered.drop(err_columns, axis=1)

# 3. Kalan saf fiziksel verilerdeki (az sayıdaki) eksik satırları tamamen sil
df_filtered = df_filtered.dropna(axis=0)

# Sonuca bakalım
df_filtered.head()
```

```python
#lineer cebirdeki gibi bir özellikler(X) ve hedef(y) matrisleri oluşturuldu.

y=df_filtered['koi_disposition'] 
X=df_filtered.drop(['koi_disposition'],axis=1)
y.head()
X.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X = X.select_dtypes(include=['number'])
X_scaled=scaler.fit_transform(X)
print("Ölçeklendirilmiş matrisin boyutu:", X_scaled.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_scaled,y,test_size=0.2,stratify=y,random_state=42)
print("Eğitim seti (Sorular):", X_train.shape)
print("Test seti (Sinav):", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression


log_model = LogisticRegression(random_state=42)

log_model.fit(X_train, y_train)

print("Model başariyla eğitildi!")
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
y_pred=log_model.predict(X_test)
acc=accuracy_score(y_test,y_pred)
print(f"modelin basari yüzdesi ={acc*100}")
confusion_matrix(y_test,y_pred)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model=RandomForestClassifier(random_state=42)
rf_model.fit(X_train,y_train)
print("RandomForest Modeli başariyla eğitildi!")
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
y_pred2=rf_model.predict(X_test)
acc2=accuracy_score(y_test,y_pred2)
print(f"modelin basari yüzdesi : {acc2*100}")
confusion_matrix(y_test,y_pred2)
```

```python
from sklearn.model_selection import cross_val_score
arr=cross_val_score(rf_model,X_scaled,y,cv=5)
x=1
sum=0
for i in arr:
    print(f"{x}'nci testin sonucu {i}")
    x=x+1
    sum=sum+i
avg=sum/5
print(f"5 testin ortalamasi:{avg}")    
```

```python
from sklearn.model_selection import GridSearchCV


param_grid={'n_estimators':[50, 100, 200],'max_depth':[None, 10, 20]}
grid_search=GridSearchCV(estimator=rf_model,param_grid=param_grid,cv=5,n_jobs=-1)
grid_search.fit(X_train, y_train)
print(grid_search.best_params_)

```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
rf_best_model=grid_search.best_estimator_
y_pred_best=rf_best_model.predict(X_test)
acc_best=accuracy_score(y_test,y_pred_best)
print(acc_best)
confusion_matrix(y_test,y_pred_best)
```

```python
import joblib


pipeline_data = {
    'model': rf_best_model,
    'scaler': scaler,
    'features': X.columns.tolist()
}


joblib.dump(pipeline_data, 'kepler_pipeline.joblib')

print("Model, Scaler ve Kolon isimleri başariyla bilgisayara kaydedildi.")
```
